```python
import networkx as nx
import pickle

with open('../pickle_files/2x3_grid_network_graph.pkl', 'rb') as f:
    network_graph = pickle.load(f)

print(network_graph.__class__)
```

```python
new_network_graph = nx.Graph()
new_network_graph.add_nodes_from(network_graph.nodes(data=True))
new_network_graph.add_edges_from(network_graph.edges(data=True))
new_network_graph.remove_edge(1, 4)

nx.draw(new_network_graph, with_labels=True)
```

```python
with open('../pickle_files/2x3_ring_network_graph.pkl', 'wb') as f:
    pickle.dump(new_network_graph, f)
```
