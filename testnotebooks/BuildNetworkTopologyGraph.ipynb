```python
import networkx as nx
import pickle
import itertools
import os

with open('../pickle_files/2x3_grid_network_graph.pkl', 'rb') as f:
    network_graph = pickle.load(f)

print(network_graph.__class__)
```

```python
for n in [6,7,8]:
    full_n_network_graph = nx.Graph()
    full_n_edges = list(itertools.combinations(range(n), 2))
    print(f"Fully connected {n}-node graph has {len(full_n_edges)} edges. : {full_n_edges}")
    full_n_network_graph.add_edges_from(full_n_edges)

    # nx.draw(full_n_network_graph, with_labels=True)
    with open(f'../networkgraphs/fully_connected_{n}_node.pkl', 'wb') as f:
        pickle.dump(full_n_network_graph, f)
```
