## 1. Verify Topology & Connectivity

Before building the circuit, we must verify:
- **Source cluster (58, 71, 77)**: Is qubit 71 connected to both 58 and 77?
- **Destination cluster (41, 53, 60)**: Is qubit 53 connected to both 41 and 60?
- **Syndrome ancillas (42, 61)**: 
  - Ancilla 42 → needs to access qubits 41 and 53 for $Z_{41}Z_{53}$ parity
  - Ancilla 61 → needs to access qubits 53 and 60 for $Z_{53}Z_{60}$ parity

````python
# Define Qubit Roles - Following Test311OnFakeBrisbane.py
# Source: 58 - 71 - 77 (data at 71, ancillas at 58 and 77)
# Destination: 41 - 53 - 60 (data arrives at 53)
# Syndrome Ancillas: 42 (for top pair 41-53), 61 (for bottom pair 53-60)

source_qubits = [58, 71, 77]  # [top, data, bottom]
dest_qubits = [41, 53, 60]    # [top', data', bottom']
ancilla_qubits = [42, 61]     # Syndrome measurement ancillas
travel_path = [59]  # Path from source to destination

def check_connection(u, v):
    """Returns True if physical wire exists between u and v."""
    return u in coupling_map.neighbors(v) or v in coupling_map.neighbors(u)

print(f"=== Checking Topology on {backend.name} ===\n")

# A. Source Check - Data qubit 71 must connect to encoding ancillas
valid_src_1 = check_connection(71, 58)
valid_src_2 = check_connection(71, 77)
print(f"SOURCE (58 - 71 - 77):")
print(f"  71-58 connected? {valid_src_1}")
print(f"  71-77 connected? {valid_src_2}")
if valid_src_1 and valid_src_2:
    print("  ✓ Source cluster is VALID")

# B. Destination Check - Center qubit 53 must connect to neighbors
valid_dst_1 = check_connection(53, 41)
valid_dst_2 = check_connection(53, 60)
print(f"\nDESTINATION (41 - 53 - 60):")
print(f"  53-41 connected? {valid_dst_1}")
print(f"  53-60 connected? {valid_dst_2}")
if valid_dst_1 and valid_dst_2:
    print("  ✓ Destination cluster is VALID")
else:
    print("  ⚠️ Destination cluster has connectivity issues!")

# C. Syndrome Ancilla Check (Critical!)
# Ancilla 42 needs to reach BOTH 41 and 53 for Z₄₁Z₅₃ parity
# Ancilla 61 needs to reach BOTH 53 and 60 for Z₅₃Z₆₀ parity
s1_check_42_41 = check_connection(42, 41)
s1_check_42_53 = check_connection(42, 53)
s2_check_61_53 = check_connection(61, 53)
s2_check_61_60 = check_connection(61, 60)

print(f"\nSYNDROME ANCILLAS:")
print(f"  Ancilla 42 (for Z₄₁Z₅₃):")
print(f"    42-41 connected? {s1_check_42_41}")
print(f"    42-53 connected? {s1_check_42_53}")
print(f"    → Direct syndrome? {s1_check_42_41 and s1_check_42_53}")

print(f"  Ancilla 61 (for Z₅₃Z₆₀):")
print(f"    61-53 connected? {s2_check_61_53}")
print(f"    61-60 connected? {s2_check_61_60}")
print(f"    → Direct syndrome? {s2_check_61_53 and s2_check_61_60}")

# Summary
print("\n=== CONNECTIVITY SUMMARY ===")
if not (s1_check_42_41 and s1_check_42_53):
    print("  ⚠️ Ancilla 42 cannot directly measure Z₄₁Z₅₃ → Need SWAP-assisted syndrome")
if not (s2_check_61_53 and s2_check_61_60):
    print("  ⚠️ Ancilla 61 cannot directly measure Z₅₃Z₆₀ → Need SWAP-assisted syndrome")
````

## 3. Build the Quantum Circuit

Following the exact logic from `Test311OnFakeBrisbane.py`, the circuit has 3 stages:

1. **Encoding** (at source 58-71-77): 
   - Prepare $|+\rangle$ on data qubit 71
   - CNOT to 58 and 77 → creates $|+\rangle_L = (|000\rangle + |111\rangle)/\sqrt{2}$

2. **Transport** (SWAP chains): 
   - Move 58 → 41 (4 SWAPs)
   - Move 71 → 53 (4 SWAPs)  
   - Move 77 → 60 (4 SWAPs)

3. **SWAP-Assisted Syndrome Measurement**:
   - Since ancillas don't directly connect to all data qubits, use **SWAP-CNOT-SWAP** pattern
   - Syndrome 1: CNOT(41→42), SWAP(53↔41), CNOT(41→42), SWAP back
   - Syndrome 2: CNOT(60→61), SWAP(60↔53), CNOT(60→61), SWAP back

````python
# Create Circuit - 2 classical bits for syndrome measurement
qc = QuantumCircuit(backend.num_qubits, 2)

# ============================================================
# STEP 1: ENCODING (at Source: 58, 71, 77)
# ============================================================
# Data qubit is 71, encoding ancillas are 58 and 77
# Prepare |+⟩ on data qubit
qc.h(71)

# Encode into [[3,1,1]] repetition code (GHZ-like state)
# |+⟩ = (|0⟩ + |1⟩)/√2  -->  (|000⟩ + |111⟩)/√2
if valid_src_1: 
    qc.cx(71, 58)  # CNOT from data to top ancilla
if valid_src_2: 
    qc.cx(71, 77)  # CNOT from data to bottom ancilla
qc.barrier(label="Encoding Complete")

print("Step 1: Encoding complete (at source 58-71-77)")
print("  - H(71) creates |+⟩")
print("  - CNOT(71→58) and CNOT(71→77) spread the state")
print("  - Logical state: (|000⟩ + |111⟩)/√2")
````

````python
# ============================================================
# STEP 2: TRANSPORT (SWAP Chains) - Following Test311OnFakeBrisbane.py
# ============================================================
# Move the encoded packet from source to destination:
#   58 → 41 (top qubit)
#   71 → 53 (data qubit)
#   77 → 60 (bottom qubit)
#
# Using the EXACT swap sequences from the test script

print("\nStep 2: Transport via SWAP chains")
print("  (Using hardcoded paths from Test311OnFakeBrisbane.py)")

# --- Move 58 → 41 (4 SWAPs) ---
print("\n  Moving top qubit 58 → 41:")
qc.swap(58, 59); print("    SWAP(58, 59)")
qc.swap(59, 60); print("    SWAP(59, 60)")
qc.swap(60, 53); print("    SWAP(60, 53)")
qc.swap(53, 41); print("    SWAP(53, 41)")

# --- Move 71 → 53 (4 SWAPs) ---
print("  Moving data qubit 71 → 53:")
qc.swap(71, 58); print("    SWAP(71, 58)")
qc.swap(58, 59); print("    SWAP(58, 59)")
qc.swap(59, 60); print("    SWAP(59, 60)")
qc.swap(60, 53); print("    SWAP(60, 53)")

# --- Move 77 → 60 (4 SWAPs) ---
print("  Moving bottom qubit 77 → 60:")
qc.swap(77, 71); print("    SWAP(77, 71)")
qc.swap(71, 58); print("    SWAP(71, 58)")
qc.swap(58, 59); print("    SWAP(58, 59)")
qc.swap(59, 60); print("    SWAP(59, 60)")

qc.barrier(label="Transport Complete")

print(f"\n  Final positions after transport:")
print(f"    • Top qubit: now at 41 (was 58)")
print(f"    • Data qubit: now at 53 (was 71)") 
print(f"    • Bottom qubit: now at 60 (was 77)")
print(f"\n  Total SWAPs for transport: 12")
````

````python
# ============================================================
# STEP 3: SWAP-ASSISTED SYNDROME MEASUREMENT
# ============================================================
# Following Test311OnFakeBrisbane.py's clever approach:
#
# Problem: Ancilla 42 connects to 41 but NOT to 53
#          Ancilla 61 connects to 60 but NOT to 53
#
# Solution: Use SWAP to temporarily bring qubits together for CNOT
#
# SYNDROME 1 (Z₄₁Z₅₃ - Top Pair):
#   1. CNOT(41 → 42)     - Link top qubit to ancilla
#   2. SWAP(53 ↔ 41)     - Bring mid qubit to position 41
#   3. CNOT(41 → 42)     - Now link mid qubit (at pos 41) to ancilla  
#   4. Measure(42 → c[0])
#   5. SWAP(41 ↔ 53)     - Restore positions
#
# SYNDROME 2 (Z₅₃Z₆₀ - Bottom Pair):
#   1. CNOT(60 → 61)     - Link bot qubit to ancilla
#   2. SWAP(60 ↔ 53)     - Bring mid qubit to position 60
#   3. CNOT(60 → 61)     - Now link mid qubit (at pos 60) to ancilla
#   4. Measure(61 → c[1])
#   5. SWAP(53 ↔ 60)     - Restore positions

print("\nStep 3: SWAP-Assisted Syndrome Measurement")
print("  (Following Test311OnFakeBrisbane.py pattern)")

# SYNDROME 1: Z₄₁Z₅₃ using ancilla 42
print("\n  Syndrome 1 (Z₄₁Z₅₃) via ancilla 42:")
qc.cx(41, 42)           # Link top qubit (now at 41) to ancilla
print("    CNOT(41 → 42) - Link top qubit")
qc.swap(53, 41)         # Swap mid qubit into position 41
print("    SWAP(53 ↔ 41) - Bring mid qubit adjacent to ancilla")
qc.cx(41, 42)           # Now link mid qubit (temporarily at 41) to ancilla
print("    CNOT(41 → 42) - Link mid qubit (now at position 41)")
qc.measure(42, 0)       # Measure syndrome bit 0
print("    Measure(42 → c[0])")
qc.swap(41, 53)         # Swap back to restore
print("    SWAP(41 ↔ 53) - Restore positions")

# SYNDROME 2: Z₅₃Z₆₀ using ancilla 61
print("\n  Syndrome 2 (Z₅₃Z₆₀) via ancilla 61:")
qc.cx(60, 61)           # Link bot qubit (now at 60) to ancilla
print("    CNOT(60 → 61) - Link bot qubit")
qc.swap(60, 53)         # Swap mid qubit into position 60
print("    SWAP(60 ↔ 53) - Bring mid qubit adjacent to ancilla")
qc.cx(60, 61)           # Now link mid qubit (temporarily at 60) to ancilla
print("    CNOT(60 → 61) - Link mid qubit (now at position 60)")
qc.measure(61, 1)       # Measure syndrome bit 1
print("    Measure(61 → c[1])")
qc.swap(53, 60)         # Swap back to restore
print("    SWAP(53 ↔ 60) - Restore positions")

qc.barrier(label="Syndrome Complete")

print("\n  === Syndrome Interpretation ===")
print("  The syndrome tells us which qubit (if any) has a bit-flip error:")
print("    00 → No error detected (state is correct)")
print("    01 → Error on qubit at position 41 (top)")
print("    10 → Error on qubit at position 60 (bottom)")
print("    11 → Error on qubit at position 53 (data/middle)")
````

## 6. Analysis: What's Going On?

### Circuit Structure (from Test311OnFakeBrisbane.py)

```
Source (58-71-77)                    Destination (41-53-60)
       58 ─────────────────────────────→ 41
       │                                  │
       71 ─────── SWAP chain ────────────→ 53  ← Data qubit
       │                                  │
       77 ─────────────────────────────→ 60

Syndrome Ancillas: 42 (measures Z₄₁Z₅₃), 61 (measures Z₅₃Z₆₀)
```

### The [[3,1,1]] Repetition Code

The **[[3,1,1]] code** encodes 1 logical qubit into 3 physical qubits:
- **Encoding**: $|0\rangle_L = |000\rangle$, $|1\rangle_L = |111\rangle$
- **Our state**: $|+\rangle_L = \frac{1}{\sqrt{2}}(|000\rangle + |111\rangle)$ (GHZ-like)

**Stabilizers** (error detection):
- $Z_1 Z_2$ → Parity of qubits 1 & 2
- $Z_2 Z_3$ → Parity of qubits 2 & 3

### The SWAP-Assisted Syndrome Trick

**Problem**: On FakeBrisbane, ancilla 42 connects to qubit 41 but NOT to 53 directly. Similarly, ancilla 61 connects to 60 but NOT to 53.

**Solution** (from the test script):
1. CNOT from the first data qubit to ancilla (direct connection)
2. SWAP the second data qubit into a position adjacent to the ancilla
3. CNOT the swapped qubit to ancilla
4. Measure the ancilla (gives XOR parity)
5. SWAP back to restore original positions

This is clever because it:
- Avoids routing the ancilla qubit around
- Only temporarily disturbs the data qubit positions
- Works within the Heavy Hex topology constraints

### Why So Many Errors?

The circuit has:
- **12 Transport SWAPs** (each = 3 CNOTs = 36 CX gates)
- **4 Syndrome SWAPs** (12 more CX gates)
- **4 Syndrome CNOTs** (4 CX gates)
- **Total**: ~52+ two-qubit gates before transpilation

After transpilation, the circuit depth grows significantly due to:
1. **Direction constraints**: CNOTs may need direction reversal (H gates)
2. **Native gate decomposition**: SWAPs → 3 CX gates each
3. **Connectivity routing**: Additional SWAPs if qubits aren't adjacent

### Expected Results

On a noisy backend like FakeBrisbane:
- **Low `00` rate**: Many errors accumulate over the deep circuit
- **Dominant error syndrome**: Often one path has worse qubits than others
- **Asymmetric distribution**: Different qubit error rates lead to non-uniform syndrome patterns

## 7. Multi-State Syndrome Analysis

Now let's systematically test different initial states for the data qubit and compare syndrome measurements. We'll test:

| State | Preparation | Encoded State |
|-------|-------------|---------------|
| $\|0\rangle$ | None (default) | $\|000\rangle$ |
| $\|1\rangle$ | X gate | $\|111\rangle$ |
| $\|+\rangle$ | H gate | $(\|000\rangle + \|111\rangle)/\sqrt{2}$ |
| $\|-\rangle$ | X + H gates | $(\|000\rangle - \|111\rangle)/\sqrt{2}$ |

For the repetition code [[3,1,1]]:
- **All four logical states** ($\|0\rangle_L$, $\|1\rangle_L$, $\|+\rangle_L$, $\|−\rangle_L$) are $+1$ eigenstates of the $Z$ stabilizers, so the ideal syndrome is always `00`.
- Any non-`00` outcome comes from transport/syndrome errors, not the encoded state.

````python
# Refactored circuit builder function
def build_311_transport_circuit_small_edge(backend, initial_state='plus', verbose=False):
    """
    Build a [[3,1,1]] encoded qubit transport circuit with syndrome measurement.
    
    Parameters:
    -----------
    backend : Backend
        The quantum backend (e.g., FakeBrisbane)
    initial_state : str
        Initial state of data qubit: 'zero', 'one', 'plus', or 'minus'
    verbose : bool
        Print circuit construction details
        
    Returns:
    --------
    QuantumCircuit
        The complete transport + syndrome circuit
    """
    # Qubit assignments (fixed for FakeBrisbane topology)
    source_qubits = [58, 71, 77]  # [top, data, bottom]
    dest_qubits = [41, 53, 60]    # [top', data', bottom']
    ancilla_qubits = [42, 61]     # Syndrome measurement ancillas
    
    data_qubit = 71  # Center qubit holds the logical data
    
    # Create circuit with 2 classical bits for syndrome
    qc = QuantumCircuit(backend.num_qubits, 2)
    
    # ============================================================
    # STEP 1: ENCODING (at Source: 58, 71, 77)
    # ============================================================
    # Prepare initial state on data qubit
    if initial_state == 'zero':
        pass  # |0⟩ is default
    elif initial_state == 'one':
        qc.x(data_qubit)  # |1⟩
    elif initial_state == 'plus':
        qc.h(data_qubit)  # |+⟩ = (|0⟩ + |1⟩)/√2
    elif initial_state == 'minus':
        qc.x(data_qubit)  # First flip to |1⟩
        qc.h(data_qubit)  # Then H gives |−⟩ = (|0⟩ - |1⟩)/√2
    else:
        raise ValueError(f"Unknown initial_state: {initial_state}")
    
    if verbose:
        print(f"Initial state: |{initial_state}⟩")
    
    # Encode into [[3,1,1]] repetition code
    # |ψ⟩ → |ψψψ⟩ (spread to all 3 qubits)
    qc.cx(data_qubit, 58)  # CNOT to top qubit
    qc.cx(data_qubit, 77)  # CNOT to bottom qubit
    qc.barrier(label="Encode")
    
    # ============================================================
    # STEP 2: TRANSPORT (SWAP Chains)
    # ============================================================
    # Move 58 → 41 (4 SWAPs)
    qc.swap(58, 59)
    qc.swap(59, 60)
    qc.swap(60, 53)
    qc.swap(53, 41)
    
    # Move 71 → 53 (4 SWAPs)
    qc.swap(71, 58)
    qc.swap(58, 59)
    qc.swap(59, 60)
    qc.swap(60, 53)
    
    # Move 77 → 60 (4 SWAPs)
    qc.swap(77, 71)
    qc.swap(71, 58)
    qc.swap(58, 59)
    qc.swap(59, 60)
    
    qc.barrier(label="Transport")
    
    # ============================================================
    # STEP 3: SWAP-ASSISTED SYNDROME MEASUREMENT
    # ============================================================
    # SYNDROME 1: Z₄₁Z₅₃ using ancilla 42
    qc.cx(41, 42)           # Link top qubit to ancilla
    qc.swap(53, 41)         # Bring mid qubit adjacent to ancilla
    qc.cx(41, 42)           # Link mid qubit to ancilla
    qc.measure(42, 0)       # Measure syndrome bit 0
    qc.swap(41, 53)         # Restore positions
    
    # SYNDROME 2: Z₅₃Z₆₀ using ancilla 61
    qc.cx(60, 61)           # Link bot qubit to ancilla
    qc.swap(60, 53)         # Bring mid qubit adjacent to ancilla
    qc.cx(60, 61)           # Link mid qubit to ancilla
    qc.measure(61, 1)       # Measure syndrome bit 1
    qc.swap(53, 60)         # Restore positions
    
    qc.barrier(label="Syndrome")
    
    return qc

# Test the function
print("Circuit builder function defined!")
print("Supported initial states: 'zero', 'one', 'plus', 'minus'")
````

````python
# Run experiments for all 4 initial states
states_to_test = ['zero', 'one', 'plus', 'minus']
state_symbols = {'zero': '|0⟩', 'one': '|1⟩', 'plus': '|+⟩', 'minus': '|−⟩'}
encoded_states = {
    'zero': '|000⟩',
    'one': '|111⟩', 
    'plus': '(|000⟩+|111⟩)/√2',
    'minus': '(|000⟩−|111⟩)/√2'
}

num_shots = 2048
results_by_state = {}

print("=" * 60)
print("Running [[3,1,1]] Transport Experiment for Different Initial States")
print("=" * 60)

for state in states_to_test:
    print(f"\n>>> Testing initial state: {state_symbols[state]} → Encoded: {encoded_states[state]}")
    
    # Build circuit
    qc = build_311_transport_circuit_small_edge(backend, initial_state=state)
    
    # Transpile
    transpiled_qc = transpile(
        qc, 
        backend, 
        initial_layout=list(range(backend.num_qubits)), 
        optimization_level=0
    )
    
    print(f"    Transpiled depth: {transpiled_qc.depth()}, CX gates: {transpiled_qc.count_ops().get('cx', 0)}")
    
    # Run simulation
    result = backend.run(transpiled_qc, shots=num_shots).result()
    counts = result.get_counts()
    
    # Store results
    results_by_state[state] = counts
    
    # Quick summary
    success_rate = counts.get('00', 0) / num_shots * 100
    print(f"    Success rate (syndrome 00): {success_rate:.1f}%")

print("\n" + "=" * 60)
print("All experiments complete!")
print("=" * 60)
````

````python
# Statistical comparison: Are the distributions significantly different?
print("=" * 70)
print("STATISTICAL ANALYSIS: Comparing Syndrome Distributions")
print("=" * 70)

# Calculate chi-square-like metric between distributions
def distribution_distance(counts1, counts2):
    """Calculate total variation distance between two distributions."""
    syndromes = ['00', '01', '10', '11']
    total1 = sum(counts1.values())
    total2 = sum(counts2.values())
    
    tvd = 0
    for s in syndromes:
        p1 = counts1.get(s, 0) / total1
        p2 = counts2.get(s, 0) / total2
        tvd += abs(p1 - p2)
    return tvd / 2  # Total Variation Distance

print("\nTotal Variation Distance (TVD) between distributions:")
print("(TVD = 0 means identical, TVD = 1 means completely different)\n")

# Pairwise comparisons
pairs = [
    ('zero', 'one'),
    ('plus', 'minus'),
    ('zero', 'plus'),
    ('one', 'plus'),
]

for s1, s2 in pairs:
    tvd = distribution_distance(results_by_state[s1], results_by_state[s2])
    print(f"  {state_symbols[s1]} vs {state_symbols[s2]}: TVD = {tvd:.4f}")

# Overall variance analysis
print("\n" + "-" * 70)
print("Success Rate (syndrome 00) Summary:")
print("-" * 70)

success_rates = []
for state in states_to_test:
    counts = results_by_state[state]
    total = sum(counts.values())
    rate = counts.get('00', 0) / total * 100
    success_rates.append(rate)
    print(f"  {state_symbols[state]}: {rate:.2f}%")

print(f"\n  Mean:   {np.mean(success_rates):.2f}%")
print(f"  Std:    {np.std(success_rates):.2f}%")
print(f"  Range:  {min(success_rates):.2f}% - {max(success_rates):.2f}%")

print("\n" + "=" * 70)
print("CONCLUSION")
print("=" * 70)
if np.std(success_rates) < 3:
    print("""
✓ The syndrome distributions are very similar across all initial states.
  This confirms that the [[3,1,1]] repetition code syndrome measurement
  is independent of the logical state - it only detects bit-flip errors,
  not the encoded information.
  
  The noise is primarily from the transport SWAPs and syndrome operations,
  not from the initial state preparation.
""")
else:
    print("""
⚠️ There is noticeable variation in syndrome distributions between states.
   This could indicate:
   - State-dependent noise (SPAM errors)
   - Asymmetric gate errors
   - Leakage or coherent errors that depend on the quantum state
""")
````
